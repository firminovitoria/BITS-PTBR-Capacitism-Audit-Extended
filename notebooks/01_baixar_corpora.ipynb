### Corpora do Hugging Face

OLID-BR, ToxSyn-PT, a parte em português do `harmful-text` (dado de treino do ToxiGuardrail-PT), HateCheck-PT e TuPy-E (versões binária e multirrótulo). Cada partição é salva em Parquet.

```python
from datasets import load_dataset

HF = {
    "olid_br":          ("dougtrajano/olid-br", None),
    "toxsyn_pt":        ("AKCIT/ToxSyn-PT", None),
    "harmful_text_pt":  ("nicholasKluge/harmful-text", "portuguese"),
    "hatecheck_pt":     ("Paul/hatecheck-portuguese", None),
    "tupye_binary":     ("Silly-Machine/TuPyE-Dataset", "binary"),
    "tupye_multilabel": ("Silly-Machine/TuPyE-Dataset", "multilabel"),
}

for name, (repo, config) in HF.items():
    ds = load_dataset(repo, config)
    for split, d in ds.items():
        d.to_parquet(f"{BASE}/{name}_{split}.parquet")
        print(name, split, d.shape)
```

### Corpora do GitHub

HateBR e HateBRXplain (trechos justificativos dos anotadores) e ToLD-Br (versão multirrótulo e partições binárias de treino, validação e teste).

```python
import urllib.request, zipfile

GH = {
    "hatebr.csv":             "https://raw.githubusercontent.com/franciellevargas/HateBR/main/dataset/HateBR.csv",
    "hatebr_xplain.csv":      "https://raw.githubusercontent.com/franciellevargas/HateBR/main/dataset/HateBRXplain.csv",
    "told_br_multilabel.csv": "https://raw.githubusercontent.com/JAugusto97/ToLD-Br/main/ToLD-BR.csv",
    "told_br_1annotator.zip": "https://github.com/JAugusto97/ToLD-Br/raw/main/experiments/data/1annotator.zip",
}

for fname, url in GH.items():
    urllib.request.urlretrieve(url, f"{BASE}/{fname}")
    print("baixado:", fname)

with zipfile.ZipFile(f"{BASE}/told_br_1annotator.zip") as z:
    z.extractall(f"{BASE}/told_br")
```

### Avaliações publicadas pelo autor do ToxiGuardrail-PT

Scores do modelo no HateCheck-PT e no teste do ToLD-Br, publicados no repositório do modelo. São usados para conferir que o modelo foi aplicado da mesma forma.

```python
from huggingface_hub import hf_hub_download
import shutil

for f in ["hatecheck_portuguese_eval.parquet", "told_br_eval.parquet"]:
    p = hf_hub_download("nicholasKluge/ToxiGuardrailPT", f)
    shutil.copy(p, f"{BASE}/toxiguardrail_{f}")
    print("baixado:", f)
```

### Conferência dos arquivos baixados

```python
import pandas as pd

for f in sorted(os.listdir(BASE)):
    path = f"{BASE}/{f}"
    if f.endswith(".parquet"):
        print(f, pd.read_parquet(path).shape)
    elif f.endswith(".csv"):
        print(f, pd.read_csv(path).shape)
```
