# 08. Subconjunto literal × figurado (RQ2)

Contribuição A. Cinquenta pares: 10 frases literais para cada termo ("cego", "surdo" e "cadeirante") e 10 figuradas para "cego" e "surdo". O rótulo (`tipo_uso`) vem da construção das frases.

- No uso **literal**, o controle remove o termo ("Meu vizinho cego tem um jardim de flores." → "Meu vizinho tem um jardim de flores.").
- No uso **figurado**, o controle troca o termo por outra palavra, registrada em `substituto` ("Ele ficou cego de raiva" → "Ele ficou tomado de raiva").

Essa diferença muda o que o delta mede e deve ser considerada na leitura (ver o notebook 09).

- **Saídas no repositório:** `data_bits/BITS_PTBR_v2_RQ2_literal_figurado_pareado.csv` e `results/BITS_PTBR_v2_literal_figurado_metricas.csv`.

**Caminhos:** este notebook foi executado no Google Colab, com os arquivos em uma pasta do Google Drive (variável `PASTA` no código). No repositório, as entradas estão em `data_bits/` e as saídas em `results/`. Para reexecutar, ajuste `PASTA`.

### Modelos (repetição da definição anterior)

```python
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

SEED = 42
N_BOOTSTRAP = 10000

np.random.seed(SEED)
torch.manual_seed(SEED)

MODELOS = {
    "BERTweet-PT": {
        "id": "pysentimiento/bertweet-pt-sentiment",
        "prefixo": "sent_bertweet",
        "tipo": "sent"
    },
    "BERTimbau Toxic": {
        "id": "dougtrajano/toxic-comment-classification",
        "prefixo": "tox1",
        "tipo": "tox1"
    },
    "ToxiGuardrail-PT": {
        "id": "nicholasKluge/ToxiGuardrailPT",
        "prefixo": "tox2",
        "tipo": "tox2"
    }
}

textos_unicos = pd.unique(
    pd.concat([
        df_bits_paired["Sentence_control"],
        df_bits_paired["Sentence"]
    ], ignore_index=True)
).tolist()

print("Dispositivo:", device)
print("Textos únicos:", len(textos_unicos))

for nome, config in MODELOS.items():
    print(nome, "=>", config["id"])
```

### Função de inferência (a mesma da v1.1)

```python
@torch.no_grad()
def inferir_modelo(
    textos,
    tokenizer,
    modelo,
    tipo,
    batch_size=32
):

    scores = []
    labels = []

    mapa = {
        str(v).upper(): int(k)
        for k, v in modelo.config.id2label.items()
    }

    if tipo == "sent":
        assert {"POS", "NEG", "NEU"}.issubset(mapa)

    if tipo == "tox1":
        assert "OFFENSIVE" in mapa

    for inicio in tqdm(
        range(0, len(textos), batch_size),
        desc=tipo
    ):

        lote = textos[inicio:inicio + batch_size]

        entradas = tokenizer(
            lote,
            padding=True,
            truncation=True,
            max_length=128,
            return_tensors="pt"
        ).to(device)

        logits = modelo(**entradas).logits

        if tipo == "sent":

            probs = torch.softmax(logits, dim=-1)

            valor = (
                probs[:, mapa["POS"]]
                - probs[:, mapa["NEG"]]
            )

            ids = probs.argmax(dim=-1).cpu().tolist()

            labels.extend([
                str(modelo.config.id2label[j]).upper()
                for j in ids
            ])

        elif tipo == "tox1":

            probs = torch.softmax(logits, dim=-1)
            valor = probs[:, mapa["OFFENSIVE"]]

        elif tipo == "tox2":

            assert logits.shape[-1] == 1, (
                "Saída inesperada do ToxiGuardrail."
            )

            valor = 1 - torch.sigmoid(
                logits.squeeze(-1)
            )

        scores.extend(
            valor.detach().cpu().numpy().tolist()
        )

    return scores, labels
```

### Inferência dos três modelos

```python
for nome, config in MODELOS.items():

    print("\n" + "=" * 60)
    print("Executando:", nome)
    print("=" * 60)

    tokenizer = AutoTokenizer.from_pretrained(
        config["id"]
    )

    modelo = AutoModelForSequenceClassification.from_pretrained(
        config["id"]
    ).to(device)

    modelo.eval()

    scores, labels = inferir_modelo(
        textos_unicos,
        tokenizer,
        modelo,
        config["tipo"],
        batch_size=32
    )

    mapa_scores = dict(zip(textos_unicos, scores))
    prefixo = config["prefixo"]

    df_bits_paired[f"{prefixo}_score_control"] = (
        df_bits_paired["Sentence_control"].map(mapa_scores)
    )

    df_bits_paired[f"{prefixo}_score_group"] = (
        df_bits_paired["Sentence"].map(mapa_scores)
    )

    df_bits_paired[f"{prefixo}_delta"] = (
        df_bits_paired[f"{prefixo}_score_group"]
        - df_bits_paired[f"{prefixo}_score_control"]
    )

    if config["tipo"] == "sent":

        mapa_labels = dict(zip(textos_unicos, labels))

        df_bits_paired["sent_control_label"] = (
            df_bits_paired["Sentence_control"].map(mapa_labels)
        )

        df_bits_paired["sent_group_label"] = (
            df_bits_paired["Sentence"].map(mapa_labels)
        )

    print("Inferência concluída:", nome)

    del tokenizer, modelo, scores, labels
    gc.collect()

    if torch.cuda.is_available():
        torch.cuda.empty_cache()

ARQUIVO_PREDICOES = (
    PASTA / "BITS_PTBR_v2_literal_figurado_com_scores.csv"
)

df_bits_paired.to_csv(
    ARQUIVO_PREDICOES,
    index=False,
    encoding="utf-8-sig"
)

print("\nPredições salvas:", ARQUIVO_PREDICOES)

display(df_bits_paired.head())
```

## 3. Métricas

### Funções das métricas (as mesmas da v1.1)

```python
def calcular_labeldist(controle, grupo, tipo):

    a = np.asarray(controle)
    b = np.asarray(grupo)

    if tipo == "sent":
        a = a == "NEG"
        b = b == "NEG"
    else:
        a = a.astype(int) == 1
        b = b.astype(int) == 1

    intersecao = int(np.sum(a & b))
    uniao = int(np.sum(a | b))

    if uniao == 0:
        return 0.0

    return float(1 - intersecao / uniao)

def calcular_fliprate(controle, grupo):

    return float(np.mean(
        np.asarray(controle) != np.asarray(grupo)
    ))

def calcular_bootstrap(delta, n_bootstrap=10000, seed=42):

    delta = np.asarray(delta, dtype=float)

    rng = np.random.default_rng(seed)

    indices = rng.integers(
        0,
        len(delta),
        size=(n_bootstrap, len(delta))
    )

    medias = delta[indices].mean(axis=1)

    li, ls = np.percentile(
        medias, [2.5, 97.5]
    )

    return float(li), float(ls)
```

### Métricas por tipo de uso e por termo

Limiar 0,5 nas métricas binárias. A versão com os limiares calibrados está no notebook 09. O ScoreDev não se aplica, porque há um único termo por frase-base.

```python
LIMIAR_TOX = 0.5

resultados = []
resultados_scoredev = []

AGRUPAMENTOS = {
    "Geral": [],
    "Por tipo": ["tipo_uso"],
    "Por termo": ["termo"],
    "Por termo e tipo": ["termo", "tipo_uso"]
}

for nome, config in MODELOS.items():

    prefixo = config["prefixo"]
    tipo = config["tipo"]

    coluna_control = f"{prefixo}_score_control"
    coluna_group = f"{prefixo}_score_group"
    coluna_delta = f"{prefixo}_delta"

    for nivel, colunas_grupo in AGRUPAMENTOS.items():

        if colunas_grupo:
            agrupados = df_bits_paired.groupby(
                colunas_grupo,
                sort=False
            )
        else:
            agrupados = [("Geral", df_bits_paired)]

        for chave, dados in agrupados:

            dados = dados.copy()

            delta = dados[coluna_delta].to_numpy(dtype=float)

            n = len(delta)
            scoresense = float(np.mean(delta))

            std = (
                float(np.std(delta, ddof=1))
                if n > 1 else np.nan
            )

            if n > 1 and std > 0:

                teste = stats.ttest_1samp(
                    delta,
                    popmean=0
                )

                t = float(teste.statistic)
                p_value = float(teste.pvalue)

            elif n > 1 and np.isclose(scoresense, 0):

                t = 0.0
                p_value = 1.0

            else:

                t = np.nan
                p_value = np.nan

            if tipo == "sent":

                labels_control = dados[
                    "sent_control_label"
                ].to_numpy()

                labels_group = dados[
                    "sent_group_label"
                ].to_numpy()

            else:

                labels_control = (
                    dados[coluna_control] >= LIMIAR_TOX
                ).astype(int).to_numpy()

                labels_group = (
                    dados[coluna_group] >= LIMIAR_TOX
                ).astype(int).to_numpy()

            labeldist = calcular_labeldist(
                labels_control,
                labels_group,
                tipo
            )

            fliprate = calcular_fliprate(
                labels_control,
                labels_group
            )

            scoredev_contexto = (
                dados.groupby("pair_id")[coluna_group]
                .agg(
                    ScoreDev_context=lambda x: float(
                        np.std(
                            x.to_numpy(dtype=float),
                            ddof=0
                        )
                    ),
                    n_termos="count"
                )
                .reset_index()
            )

            scoredev = float(
                scoredev_contexto["ScoreDev_context"].mean()
            )

            scoredev_contexto["Modelo"] = nome
            scoredev_contexto["Nivel"] = nivel

            resultados_scoredev.append(scoredev_contexto)

            li, ls = calcular_bootstrap(
                delta,
                N_BOOTSTRAP,
                SEED
            )

            if not isinstance(chave, tuple):
                chave = (chave,)

            termo = "Todos"
            tipo_uso = "Todos"

            if nivel == "Por tipo":
                tipo_uso = chave[0]

            elif nivel == "Por termo":
                termo = chave[0]

            elif nivel == "Por termo e tipo":
                termo, tipo_uso = chave

            resultados.append({
                "Modelo": nome,
                "Nivel": nivel,
                "Termo": termo,
                "tipo_uso": tipo_uso,
                "ScoreSense": scoresense,
                "Std": std,
                "t": t,
                "p-value": p_value,
                "n": n,
                "LabelDist": labeldist,
                "FlipRate": fliprate,
                "ScoreDev": scoredev,
                "IC95_inf": li,
                "IC95_sup": ls
            })

df_metricas = pd.DataFrame(resultados)

df_scoredev_contextos = pd.concat(
    resultados_scoredev,
    ignore_index=True
)

print("MÉTRICAS GERAIS")
display(
    df_metricas[
        df_metricas["Nivel"] == "Geral"
    ].round(5)
)

print("\nMÉTRICAS POR TIPO DE USO")
display(
    df_metricas[
        df_metricas["Nivel"] == "Por tipo"
    ].round(5)
)

print("\nMÉTRICAS POR TERMO E TIPO")
display(
    df_metricas[
        df_metricas["Nivel"] == "Por termo e tipo"
    ].round(5)
)
```

### Comparação literal × figurado

```python
comparacoes = []

for nome, config in MODELOS.items():

    coluna = f"{config['prefixo']}_delta"

    for termo in ["cego", "surdo"]:

        dados = df_bits_paired[
            df_bits_paired["termo"] == termo
        ]

        literal = dados.loc[
            dados["tipo_uso"] == "literal",
            coluna
        ].to_numpy(dtype=float)

        figurado = dados.loc[
            dados["tipo_uso"] == "figurado",
            coluna
        ].to_numpy(dtype=float)

        teste = stats.ttest_ind(
            literal,
            figurado,
            equal_var=False
        )

        rng = np.random.default_rng(SEED)

        boot_literal = rng.choice(
            literal,
            size=(N_BOOTSTRAP, len(literal)),
            replace=True
        ).mean(axis=1)

        boot_figurado = rng.choice(
            figurado,
            size=(N_BOOTSTRAP, len(figurado)),
            replace=True
        ).mean(axis=1)

        diferencas = boot_literal - boot_figurado

        li, ls = np.percentile(
            diferencas,
            [2.5, 97.5]
        )

        comparacoes.append({
            "Modelo": nome,
            "Termo": termo,
            "n_literal": len(literal),
            "n_figurado": len(figurado),
            "Media_literal": literal.mean(),
            "Media_figurado": figurado.mean(),
            "Diferenca_literal_menos_figurado":
                literal.mean() - figurado.mean(),
            "t_Welch": teste.statistic,
            "p-value": teste.pvalue,
            "IC95_inf": li,
            "IC95_sup": ls
        })

df_comparacao = pd.DataFrame(comparacoes)

display(df_comparacao.round(5))
```
